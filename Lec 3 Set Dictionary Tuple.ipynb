# Set
Unordered collection data type that is 
Iterable ,
Mutable ,
No duplicate elements , no indexing ,
Based on hash table data structure


```python
st=set()
print(st)
print(type(st))
```

```python
st={1,2,3,44,5,6,5}
st
```

```python
st={"hey", "there" , "Stop " , "Stalking" , "my", "Git"}
st
```

```python
st.add("BB")
st
```

```python
st2={"gotcha" , 'never' , 'reached' , 'there' , 'BB'}
st2.difference(st)
st.difference(st2)
st.union(st2)
st.intersection(st2)
```

```python
print(st)
st.difference_update(st2)
st
```

# Dictionary
It is a collection which is unordered , changeable and indexed . In python dictionaried are written with curly brackets, and have keys and values

```python
dict={}
print(type(dict))
dict={1,2,3,4,5,6,7}
print (dict)
print(type(dict))
```

```python
d={1:"This" , 2:"is" , 3:"the " , 4:"end"}
print(type(d))
for c in d:
    print (c)
```

```python
for c in d.values():
    print(c)
    print(d)
print(d)
```

# Nested dictionary


```python
car1_model={'Mercedes':1980}
car2_model={'Audi':1970}
car3_model={'Ambassador':1980}
car_type={'car1':car1_model,'car2':car2_model,'car3':car3_model}
print(car_type)
print(car_type["car1"])
```

```python
for c in car_type.values():
    print(c)
```

```python
print(car_type['car1']['Mercedes'])
```
